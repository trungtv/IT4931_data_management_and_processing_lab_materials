# Multipart và snapshot hiện tại

Part ở giữa phải từ 5MiB, không quá 10000 part. Object nhỏ là một part. `ListObjects` không cho biết snapshot Iceberg nào đang current. Lời giải: `S3_lab/exercises/layout.py`.

```python

import sys
from pathlib import Path

def add_exercises(relative):
    here = Path.cwd().resolve()
    for candidate in [here, *here.parents]:
        folder = candidate / relative
        if folder.is_dir():
            sys.path.insert(0, str(folder))
            return
    raise RuntimeError(f"Không thấy {relative}")

add_exercises("S3_lab/exercises")
from layout import current_snapshot_readable_by, metadata_key, multipart_plan
five = 5 * 1024 * 1024
print(multipart_plan(five + 10, five))
print(metadata_key("silver_trades", 7))
print(current_snapshot_readable_by("list_objects"))
```
