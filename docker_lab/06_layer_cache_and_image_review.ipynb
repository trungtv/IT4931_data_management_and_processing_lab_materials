# Layer cache và Dockerfile

Đổi một instruction thì mọi layer phía sau phải build lại. Image của lab Airflow được ghim tag, không dùng ADD, không chạy root ở layer cuối, và xóa cache apt/pip. Lời giải: `docker_lab/exercises/image_review.py`.

```python

import sys
from pathlib import Path

def add_exercises(relative):
    here = Path.cwd().resolve()
    for candidate in [here, *here.parents]:
        folder = candidate / relative
        if folder.is_dir():
            sys.path.insert(0, str(folder))
            return
    raise RuntimeError(f"Không thấy {relative}")

add_exercises("docker_lab/exercises")
from pathlib import Path
from image_review import layers_rebuilt, review_dockerfile
before = ["FROM python:3.11", "COPY requirements.txt .", "RUN pip install -r requirements.txt", "COPY src src"]
after = ["FROM python:3.11", "COPY requirements.txt .", "RUN pip install -r requirements.txt", "COPY app app"]
print(layers_rebuilt(before, after))
here = Path.cwd().resolve()
for candidate in [here, *here.parents]:
    dockerfile = candidate / "platform" / "airflow" / "Dockerfile"
    if dockerfile.is_file():
        print(review_dockerfile(dockerfile.read_text()))
        break
else:
    raise RuntimeError("Không thấy platform/airflow/Dockerfile")
```
