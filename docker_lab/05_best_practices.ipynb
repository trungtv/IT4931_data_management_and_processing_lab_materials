# Lab 5: Soát image đang dùng

Dockerfile sạch và container thật cùng nằm trên network `it4931`.

```python
import sys
from pathlib import Path
here = Path.cwd().resolve()
for candidate in [here, *here.parents]:
    lab = candidate / "docker_lab" / "lab"
    compose = candidate / "platform" / "docker-compose.yml"
    dockerfile = candidate / "platform" / "airflow" / "Dockerfile"
    if lab.is_dir() and compose.is_file():
        sys.path.insert(0, str(lab))
        break
else:
    raise RuntimeError("Không thấy platform")

from platform_inspect import review_airflow_image, running_airflow
print(review_airflow_image(dockerfile))
print(running_airflow()['networks'])
```
