# Lab 1: Nền Airflow đang chạy

API server là `airflow-apiserver`, cổng 8080. Health của Airflow 3.1 là `/api/v2/monitor/health`, không phải `/health`. CLI chạy trong `airflow-scheduler`. DAG của bài này là `trade_quality_gate`.

```python
import sys
from pathlib import Path
here = Path.cwd().resolve()
for candidate in [here, *here.parents]:
    exercises = candidate / "Airflow_lab" / "exercises"
    pipeline = candidate / "Data_Lakehouse_lab" / "pipeline"
    if exercises.is_dir() and pipeline.is_dir():
        sys.path.insert(0, str(pipeline))
        sys.path.insert(0, str(exercises))
        break
else:
    raise RuntimeError("Không thấy Airflow_lab/exercises")

import json
import urllib.request
with urllib.request.urlopen("http://localhost:8080/api/v2/monitor/health", timeout=5) as response:
    print(response.status, json.load(response)["metadatabase"]["status"])
```
