# Lab 8: Retry và schema

`broker_down` được chạy lại cho đến hết số lần. `schema` dừng ngay. Đó là lý do task `publish` đổi `ValueError` thành `AirflowFailException`.

```python
import sys
from pathlib import Path
here = Path.cwd().resolve()
for candidate in [here, *here.parents]:
    exercises = candidate / "Airflow_lab" / "exercises"
    pipeline = candidate / "Data_Lakehouse_lab" / "pipeline"
    if exercises.is_dir() and pipeline.is_dir():
        sys.path.insert(0, str(pipeline))
        sys.path.insert(0, str(exercises))
        break
else:
    raise RuntimeError("Không thấy Airflow_lab/exercises")

from idempotency import retry_decision
print(retry_decision("schema", 1, 3), retry_decision("broker_down", 1, 3), retry_decision("broker_down", 3, 3))
```
