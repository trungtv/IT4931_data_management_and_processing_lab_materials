# Chẩn đoán consumer

Service chọn nhầm label thì không có endpoint. Request lớn hơn limit bị từ chối. Limit nhỏ hơn bộ nhớ process cần thì OOM. Probe TCP xanh trong khi tiến trình chỉ sẵn sàng khi HTTP health trả lời. Lời giải: `K8s_Basics_lab/exercises/diagnose.py`.

```python

import sys
from pathlib import Path

def add_exercises(relative):
    here = Path.cwd().resolve()
    for candidate in [here, *here.parents]:
        folder = candidate / relative
        if folder.is_dir():
            sys.path.insert(0, str(folder))
            return
    raise RuntimeError(f"Không thấy {relative}")

add_exercises("K8s_Basics_lab/exercises")
from diagnose import diagnose_workload
print(diagnose_workload({
    "pod_labels": {"app": "trade-consumer"},
    "service_selector": {"app": "trades"},
    "container_port": 8080,
    "readiness_port": 8080,
    "memory_request_mi": 256,
    "memory_limit_mi": 128,
    "memory_needed_mi": 400,
    "probe": "tcp",
    "ready_when": "http_health",
}))
```
