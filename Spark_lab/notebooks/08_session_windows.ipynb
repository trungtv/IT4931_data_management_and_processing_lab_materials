# Session window

Hai giao dịch thuộc cùng một session khi khoảng cách không lớn hơn `gap`. Đầu vào không được giả định đã sắp. Lời giải: `Spark_lab/exercises/sessions.py`.

```python

import sys
from pathlib import Path

def add_exercises(relative):
    here = Path.cwd().resolve()
    for candidate in [here, *here.parents]:
        folder = candidate / relative
        if folder.is_dir():
            sys.path.insert(0, str(folder))
            return
    raise RuntimeError(f"Không thấy {relative}")

add_exercises("Spark_lab/exercises")
import pandas as pd
from sessions import sessionize
print(sessionize([pd.Timestamp("2026-11-17 09:10"), pd.Timestamp("2026-11-17 09:00"), pd.Timestamp("2026-11-17 09:04")], pd.Timedelta(minutes=5)))
```
