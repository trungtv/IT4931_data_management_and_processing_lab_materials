# Suite giao dịch trên GX 1.x

Context ephemeral, không dùng `gx.from_pandas`. Batch fail khi `trade_id` null hoặc trùng, `side` khác BUY/SELL, hoặc `price` không dương. Lời giải: `Great_Expectations_lab/exercises/trade_suite.py`.

```python

import sys
from pathlib import Path

def add_exercises(relative):
    here = Path.cwd().resolve()
    for candidate in [here, *here.parents]:
        folder = candidate / relative
        if folder.is_dir():
            sys.path.insert(0, str(folder))
            return
    raise RuntimeError(f"Không thấy {relative}")

add_exercises("Great_Expectations_lab/exercises")
import pandas as pd
from trade_suite import failed_expectations
clean = pd.DataFrame([{"trade_id": "T1", "side": "BUY", "price": 22.0}, {"trade_id": "T2", "side": "SELL", "price": 21.5}])
dirty = pd.DataFrame([{"trade_id": "T1", "side": "BUY", "price": 22.0}, {"trade_id": "T1", "side": "HOLD", "price": 0}])
print(failed_expectations(clean))
print(failed_expectations(dirty))
```
