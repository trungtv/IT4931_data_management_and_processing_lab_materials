# Lab 3: Checkpoint dừng batch lỗi

Checkpoint không chỉ in kết quả. Còn expectation fail thì ném lỗi và batch không được đi tiếp.

```python
import sys
from pathlib import Path
here = Path.cwd().resolve()
for candidate in [here, *here.parents]:
    folder = candidate / "Great_Expectations_lab" / "exercises"
    if folder.is_dir():
        sys.path.insert(0, str(folder))
        break
else:
    raise RuntimeError("Không thấy Great_Expectations_lab/exercises")
import pandas as pd
from trade_suite import checkpoint, failed_expectations
clean = pd.DataFrame([
    {"trade_id": "T1", "side": "BUY", "price": 22.0},
    {"trade_id": "T2", "side": "SELL", "price": 21.5},
])
dirty = pd.DataFrame([
    {"trade_id": "T1", "side": "BUY", "price": 22.0},
    {"trade_id": "T1", "side": "HOLD", "price": 0},
])

checkpoint(clean)
try:
    checkpoint(dirty)
except RuntimeError as error:
    print(error)
```
