# Quyết định compaction và time travel

File nhỏ được gom đến khi nhóm chạm `target_bytes`. File đã đủ lớn không bị viết lại. Time travel lấy snapshot mới nhất có `committed_at <= as_of`; hai snapshot cùng thời điểm thì lấy id lớn hơn.

Lời giải nằm trong `pyiceberg/exercises/plan.py`.

```python

import sys
from pathlib import Path

def add_exercises(relative):
    here = Path.cwd().resolve()
    for candidate in [here, *here.parents]:
        folder = candidate / relative
        if folder.is_dir():
            sys.path.insert(0, str(folder))
            return
    raise RuntimeError(f"Không thấy {relative}")

add_exercises("pyiceberg/exercises")
from datetime import datetime
from plan import plan_compaction, snapshot_as_of

print(plan_compaction([("a", 40), ("b", 40), ("c", 30), ("big", 100)], 100))
print(snapshot_as_of([(1, datetime(2026,11,17,9,0)), (4, datetime(2026,11,17,10,0)), (3, datetime(2026,11,17,10,0))], datetime(2026,11,17,10,0)))
```
