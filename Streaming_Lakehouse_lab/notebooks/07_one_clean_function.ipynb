# Một hàm clean cho batch và stream

Dọn từng micro-batch rồi nối lại vẫn để lọt `trade_id` xuất hiện ở hai batch. Cùng `clean_trades`, chạy một lần trên toàn bộ dòng, thì khóa đó còn một bản. Lời giải: `Streaming_Lakehouse_lab/exercises/same_function.py`.

```python

import sys
from pathlib import Path

def add_exercises(relative):
    here = Path.cwd().resolve()
    for candidate in [here, *here.parents]:
        folder = candidate / relative
        if folder.is_dir():
            sys.path.insert(0, str(folder))
            return
    raise RuntimeError(f"Không thấy {relative}")

add_exercises("Streaming_Lakehouse_lab/exercises")
add_exercises("Data_Lakehouse_lab/pipeline")
import pandas as pd
from same_function import clean_all_at_once, duplicate_ids_when_cleaned_per_batch
from transforms import clean_trades

def clean(frame):
    return clean_trades(frame, as_of=pd.Timestamp("2026-11-17 12:00:00+00:00"))

row = lambda trade_id: {"trade_id": trade_id, "symbol": "fpt", "side": "buy", "price": 22, "quantity": 10, "event_time": "2026-11-17 09:00:00+00:00"}
frames = [pd.DataFrame([row("T1"), row("T2")]), pd.DataFrame([row("T1")])]
print(duplicate_ids_when_cleaned_per_batch(frames, clean))
print(clean_all_at_once(frames, clean)["trade_id"].tolist())
```
