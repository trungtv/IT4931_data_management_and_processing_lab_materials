# Merge incremental theo trade_id

Bản incoming thắng khi `event_time` mới hơn hoặc bằng bản đang có. Bản cũ hơn không được ghi đè. Đây là hợp đồng của incremental model, kiểm bằng pytest trước khi viết SQL. Lời giải: `dbt_lab/exercises/merge_trades.py`.

```python

import sys
from pathlib import Path

def add_exercises(relative):
    here = Path.cwd().resolve()
    for candidate in [here, *here.parents]:
        folder = candidate / relative
        if folder.is_dir():
            sys.path.insert(0, str(folder))
            return
    raise RuntimeError(f"Không thấy {relative}")

add_exercises("dbt_lab/exercises")
import pandas as pd
from merge_trades import merge_trades
existing = pd.DataFrame([{"trade_id":"T1","symbol":"FPT","side":"BUY","price":20,"quantity":10,"event_time":"2026-11-17 09:00:00+00:00"}])
incoming = pd.DataFrame([{"trade_id":"T1","symbol":"FPT","side":"BUY","price":22,"quantity":10,"event_time":"2026-11-17 09:05:00+00:00"}])
print(merge_trades(existing, incoming))
```
