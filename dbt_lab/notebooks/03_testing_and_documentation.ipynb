# Lab 3: Test

`schema.yml` bắt `symbol` not null và unique, `trade_count` not null. `dbt build` ở lab 1 chạy các test này.

```python
import sys
from pathlib import Path
here = Path.cwd().resolve()
for candidate in [here, *here.parents]:
    project = candidate / "dbt_lab" / "trade_project"
    if project.is_dir():
        break
else:
    raise RuntimeError("Không thấy trade_project")

text = (project / "models" / "schema.yml").read_text()
for name in ("not_null", "unique", "trade_count"):
    if name not in text:
        raise AssertionError(name)
print(text)
```
