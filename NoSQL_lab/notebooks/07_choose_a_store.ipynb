# Chọn store theo hình dạng câu hỏi

Giá mới nhất có TTL là Redis. Hồ sơ một giao dịch là MongoDB. Từ hai cạnh trở lên là Neo4j, kể cả khi pattern bị ghi là document. SQL bắc qua nhiều store là Trino. Đọc điểm không có TTL thì không chọn bừa. Lời giải: `NoSQL_lab/exercises/model_choice.py`.

```python

import sys
from pathlib import Path

def add_exercises(relative):
    here = Path.cwd().resolve()
    for candidate in [here, *here.parents]:
        folder = candidate / relative
        if folder.is_dir():
            sys.path.insert(0, str(folder))
            return
    raise RuntimeError(f"Không thấy {relative}")

add_exercises("NoSQL_lab/exercises")
from model_choice import QueryNeed, choose_store
print(choose_store(QueryNeed("point", hops=0, ttl_seconds=30)))
print(choose_store(QueryNeed("document", hops=2)))
print(choose_store(QueryNeed("federate", hops=0)))
```
